The code cell below runs in the browser when the dashboard opens. If the kernel starts, it prints its Python version and shows a widget.

```python
import sys, platform
import ipywidgets as w
print(platform.python_implementation(), sys.version.split()[0], sys.platform)
slider = w.IntSlider(value=3, min=0, max=10, description='n')
label = w.Label()
def show(change=None):
    label.value = f'n squared = {slider.value ** 2}'
slider.observe(show, 'value'); show()
w.VBox([slider, label])
```
