```python
import inspect
import platform
import sys

import ipywidgets as w


def demo():
    """The widgets of this dashboard."""
    slider = w.IntSlider(value=3, min=0, max=10, description="n")
    label = w.Label()

    def show(change=None):
        label.value = f"n squared = {slider.value ** 2}"

    slider.observe(show, "value")
    show()
    return w.VBox([slider, label])
```

```python
print(inspect.getsource(demo))
```
